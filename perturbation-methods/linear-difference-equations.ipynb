# Linear Difference Equations

In this lab, we'll discuss how to solve systems of linear difference equations, compute the stationary distribution of a Markov chain and simulate a Markov chain using Monte Carlo methods (MCMC).

We'll see that the solution to a linear difference equation is inherently connected to the eigenvalues of the matrix representing the system. Therefore, we'll also need to discuss how to compute eigenvalues numerically.

As a first example, lets consider the following linear difference equation system. Let $y_t \in \mathbb{R}$

\begin{align}
    y_t=\rho y_{t-1}
\end{align}

What's the solution to the system? Since there no other variables on the right hand side, the easiest approach is to iteratively substitute for $y_{t-1}$. This gives us 

\begin{align}
    y_t=\rho^{t}y_{0}
\end{align}

where $y_0$ is given. If $|\rho|<1$, then the solution is finite. If $|\rho|>1$, the solution is infinite. Note that $\rho$ is the only eigenvalue of the matrix representing the system. The matrix $\left[\rho\right]$ is diagonal. As you know, the eigenvalues of a diagonal matrix are the elements on the diagonal. Therefore, in the one dimensional case, we see that there is a strong connection between the eigenvalues of the system and the solution to the difference equation.



Now, lets consider a $2 \times 2$ system. 

\begin{align}
    \begin{bmatrix}
        y_{1,t} \\
        y_{2,t} 
    \end{bmatrix} =
     \begin{bmatrix}
        a_{1,1} & 0 \\
        0 & a_{2,2} 
    \end{bmatrix}
    \begin{bmatrix}
        y_{1,t-1} \\
        y_{2,t-1} 
    \end{bmatrix}
\end{align}


The matrix is diagonal. Therefore, we can solve each equation as a univariate equation. Using this insight, we see that the solution is going to be 
\begin{align}
    \begin{bmatrix}
        y_{1,t} \\
        y_{2,t} 
    \end{bmatrix} =
     \begin{bmatrix}
        a_{1,1}^t & 0 \\
        0 & a_{2,2}^t 
    \end{bmatrix}
    \begin{bmatrix}
        y_{1,0} \\
        y_{2,0} 
    \end{bmatrix}
\end{align}

Notice that we can rewrite the matrix as 
\begin{align}
\begin{bmatrix}
        a_{1,1}^t & 0 \\
        0 & a_{2,2}^t 
    \end{bmatrix}=
    \begin{bmatrix}
        a_{1,1} & 0 \\
        0 & a_{2,2} 
    \end{bmatrix}^t
\end{align}

In other words, if the matrix for the system is $A$, the solution is $A^t$ times the initial conditions. Again, since our matrix is diagonal, the eigenvalues of the system matrix are just the diagonal elements. Therefore, we see that even for $2 \times 2$ diagonal matricies, the solution is inherently connected to the eigenvalues of the system. 



Now, lets consider the $n \times n$ system where $A$ is not diagonal. However, we'll assume that $A$ is diagonalizable.

\begin{align}
    Y_t=A Y_{t-1}
\end{align}

Since $A$ is diagonalizable, we can write $A$ as $PDP^{-1}$. Let $Z_t\triangleq P^{-1} Y_t$. Then, we can write the system as 

\begin{align}
    Z_t=D Z_{t-1}
\end{align}

Since $D$ is diagonal, we have a system of $n$ univariate equations. We know that the solution to this system is 

\begin{align}
    Z_t=\begin{bmatrix}
    \lambda_1^t & 0 &\cdots & 0 &0 \\
    0 & \lambda_2^t &\cdots & 0 &0 \\
    \vdots & & & & \vdots \\
    0 & 0 &\cdots & \lambda_{n-1}^t &0 \\
    0 & 0 &\cdots & 0 &\lambda_n^t \\
    \end{bmatrix} Z_0 =D^tZ_0
\end{align}


We can then recover the original system by multiplying both sides by $P$.

\begin{align}
    Y_t=PD^t P^{-1} Y_0
\end{align}

Now, we can see that if $|\lambda_i| >1$, then $Y_t$ may contain values going off to infinity. Suppose that we want all of our values to be finite. How can we ensure this? What we can do is choose $Y_0$ in such a way to ensure this. In most economic examples, only a subset of the variables have initial conditions given. For example, in the basic model seen in class, only $k_{-1}$ is given, but $c_{-1}$ is not. Therefore, $c_{-1}$ is a a free parameter that we can use to ensure that the solution doesn't explode and violate the transversality condition. As you'll discuss in class, this is related the Blanchard-Kahn conditions.

Let's see exactly how we can do this since we need to be concrete when we go to Matlab. Specifically, how do we ensure that our system is non explosive? Note the following 

\begin{align}
    PD^tZ_0=&\left(   
    \lambda_1^t\begin{bmatrix}
    P_1 & 0 &\cdots & 0 &0 
    \end{bmatrix}Z_0 + \lambda_1^t\begin{bmatrix}
    0 & P_2 &\cdots & 0 &0 
    \end{bmatrix}Z_0 +\lambda_{n-1}^t\begin{bmatrix}
    0 & 0 &\cdots & P_{n-1} &0 
    \end{bmatrix}Z_0+\lambda_n^t\begin{bmatrix}
    0 & 0 &\cdots & 0 &P_n
    \end{bmatrix}Z_0 \right) \\
    =& \sum_{i=1}^N \lambda_i^t P_i z_{i,0}
\end{align}

We know that we're choosing $Y_0$ in such a way that $z_{i,0}=0$ if $|\lambda_i|>1$. Therefore, when we actually implement the code, we know that we can transform the $P$ matrix so that all eigenvectors corresponding to eigenvalues with $|\lambda_i|>1$ are set to zero. That is, if only $|\lambda_1|$ and $|\lambda_2|$ are less than 1, we can create a matrix called $\tilde{P}$ where 
 
 \begin{align}
      \tilde{P}=\begin{bmatrix}
    P_1 & P_2 & 0 &\cdots & 0 &0 
    \end{bmatrix}
 \end{align}
 
 Additionally, we can create a matrix called $\tilde{D}$ which is a transformation of $D$ where all eigenvalues with absolute value greater than unity are set to zero.
 
 \begin{align}
    \tilde{D}=\begin{bmatrix}
    \lambda_1 & 0 &\cdots & 0 &0 \\
    0 & \lambda_2 &\cdots & 0 &0 \\
    \vdots & & & & \vdots \\
    0 & 0 &\cdots & 0 &0 \\
    0 & 0 &\cdots & 0 & 0 \\
    \end{bmatrix}
\end{align}

For computational purposes, its more efficient to only store the columns of $\tilde{P}$ that are non-zero and the block of $\tilde{D}$ associated with non-zero entries (Footnote: if we stored $\tilde{P}$ and $\tilde{D}$ as a sparse matrix there may be little efficiency gain). Next, we construct $\tilde{Z_0}$ where again, only the entries associated with eigenvalues smaller than unity are stored. In our example, $\tilde{D}_{new}$ is $2 \times 2$, $\tilde{P}_{new}$ is $n \times 2$ and $Z_{0,new}$ is $2 \times 1$. With this we can iterate the system forward 

\begin{align}
    Z_{1,new}=\tilde{D}_{new}Z_{0,new} \\
    Y_1=\tilde{P}_{new}Z_{1,new}
\end{align}

Now lets look at the code. The main functions we'll need are $\mathtt{eig}$ and $\mathtt{sort}$. We'll also need to know how to reorder the columns of a matrix based on an index.

```python
clear all
c=-2;
b=1;
d=(-1+sqrt(1-4*(-3/4+c*b)))/2;
a=-1-d;
A=[a b; c d];
Y0=[1.5;0.5];

[nr,nc]=size(A);
```

The function $\mathtt{eig}$ when we specify we want two returns provides us the eigenvalues and the eigenvectors. The eigenvalues are not necessarily sorted. Therefore, we need to use the $\mathtt{sort}$ function. $B$ returns the sorted vector and $I$ returns the index used to sort the vector.

The function $\mathtt{diag}$, when provided a matrix as input, returns the elements from the diagonal. If you provide the function $\mathtt{diag}$ a vector, it returns a matrix with the elements of the vector on the diagonal.

```python
% Get eigenvalues and Eigenvectors
[V,D]=eig(A);
[B,I] = sort(abs(diag(D)));
D_sort=D(I,I);
D_new=diag(D_sort(B<1.0));
N_exp_ev=sum(B>1.0);
V_star=V(:,I(nc-N_exp_ev));
```

Finally, we need to set our initial conditions. First we construct $Z_0$. Then, we extract the $z_{i,0}$'s associated with non-explosive eigenvalues.

```python
% Set initial conditions 

Z0=V\Y0;
Z0_new=Z0(I(nc-N_exp_ev));


T=50;
Z=zeros(nc-N_exp_ev,T);
Z(:,1)=Z0_new;
Y=zeros(nc,T);
for t=2:T 
    Z(:,t)=D_new*Z(:,t-1);
    Y(:,t)=V_star*Z(:,t);
end
```

## Markov Chains
Let us recall some facts about Markov chains. The first thing to recall is the definition of an ergodic Markov chain.

**Definition** A stationary measure is a measure $\mu$ such that 
\begin{align}
    \mu\left(j\right)=\sum_{i=1}^N \pi_{i,j} \mu\left(i\right)  & & \forall j
\end{align}

Using the Notation from class, we can rewrite this in matrix notation as 

\begin{align}
    \mu=\Pi^{\prime} \mu
\end{align}
Therefore, we have 
\begin{align}
    \left(1-\Pi^{\prime}\right)\mu=0
\end{align}

In other words, the stationary distribution is the eigenvector associated with the unit eigenvalue of $\Pi$. Note that since $\Pi$ is a stochastic matrix, $\Pi^{\prime}$ will always have at least one unit eigenvalue. It is also possible for the matrix to have more than one unit eigenvalue. In this case, the stationary distribution is indeterminate. 
There are conditions that we can check to ensure that the we will have a unique invariant stationary distribution. 

**Condition 1**: If $\Pi_{i,j}>0$ $\forall i,j$ then $\Pi$ has a unique stationary distribution and the process is asymptotically stationary. 

**Condition 2**: If $\Pi^n_{i,j}>0$ $\forall i,j$ for some $n\geq1$ then $\Pi$ has a unique stationary distribution and the process is asymptotically stationary.


## Simulation
The next question is is how do we go about simulating a Markov chain. Let, the current state be denoted as $s_t$. Notice that each row of the transition matrix $\Pi$ is a conditional distribution. That is, row $i$ tells us the probability of moving to state $j\in \mathcal{S}$ conditional on being in state $i$ today. Since each row sums to one, and the columns of each row are mutually exclusive, we can see that this indeed a probability distribution. \\
The reason this is useful is that we can simulate the Markov chain using a random number generator where the random variable is distributed uniform on the unit interval. To see this consider row $i$ of $\Pi$
\begin{align}
\begin{bmatrix}
    \pi_{i,1} & \pi_{i,2} & \pi_{i,2} & ... & \pi_{i,N}
\end{bmatrix}
\end{align}

Since the probability of a uniformly distributed random variable on the unit interval falling inside of a small interval of size $\delta$ is the same regardless of there the interval (so long as it is on the support), we can split the unit interval in a way so that we can achieve the desired probabilities. 

Specifically, we proceed as follows. Let $U_{1;i}=\left(0,\pi_{i,1}\right]$ and $U_{j;i}=\left(\sum_{k=1}^{j-1} \pi_{i,k},\sum_{k=1}^{j}\pi_{i,k}\right]$. Now, consider a realization of the random variable $X$, where $X \sim \mathcal{U}\left(0,1\right)$. If $x\in U_{j;i}$, then $s_{t+1}=j$. Notice that $\pi_{i,j}=Pr\left(s_{t+1}=j;s_t=i\right)=\mu\left(U_j\right)=Pr\left(x\in U_{j;i}\right)$. By using this partition, we can simulate the Markov chain. 

The algorithm is roughly as follows:
1. Initialize $s_0=i$
2. Draw $x_n$ and set $s_n=j$ if $x_n\in U_{j;i}$
3. Set $i=j$ and return to step 2. Repeat $N$ times.

Note that in step 3, when setting $i=j$ we are changing the conditional probabilty distribution that we are using. In other words, if we set $s_0=3$ we need the distribution induced by $\left\{ U_{j;3}\right\}$.If $s_1=4$, then we need $\left\{ U_{j;4}\right\}$. Setting $i=j$ in step 3 produces this result. 

## Example

Let's consider a model of employment. The transitions in and out of employment are governed by the following transition matrix.

\begin{align}
    P=\begin{bmatrix}
    p_{u,u} & p_{u,e} \\
    p_{e,u} & p_{e,e} 
    \end{bmatrix}
\end{align}

Suppose the agent initially starts out employed. Now, we need to draw from the uniform distribution. Denote our draw as $x_1$. If $x_1\leq p_{e,u}$, then the agent becomes unemployed. Otherwise, the agent remains employed.

```python
clear all
p_uu=0.4;
p_ue=1.0-p_uu;
p_eu=0.2;
p_ee=1.0-p_eu;

P=[ p_uu p_ue; p_eu p_ee];
T=500000;
% Initialize state
s=zeros(1,T);
% Initialize the agent to being employed
s(1)=2;

for i=2:T
    x=rand();
    if x<=P(s(i-1),1)
        s(i)=1;
    else
        s(i)=2;
    end
end

e_rate=sum(s==1)/T;
u_rate=sum(s==2)/T;

[V,D]=eig(P');

unit_ev_loc=find(diag(D)==1);
stat_dist=abs(V(:,unit_ev_loc))/sum(abs(V(:,unit_ev_loc)));

norm([e_rate; u_rate]-stat_dist)
```
